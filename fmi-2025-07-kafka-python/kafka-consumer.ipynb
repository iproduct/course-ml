```python
import msgpack
from kafka import KafkaConsumer

consumer = KafkaConsumer('foobar')
for msg in consumer:
    print(msg)
```

```python
# join a consumer group for dynamic partition assignment and offset commits
from kafka import KafkaConsumer

consumer = KafkaConsumer('my_favorite_topic', group_id='my_favorite_group')
for msg in consumer:
    print(msg)
```

```python
# manually assign the partition list for the consumer
from kafka import TopicPartition

consumer = KafkaConsumer(bootstrap_servers='localhost:9092')
consumer.assign([TopicPartition('events', 0)])
msg = next(consumer)
print(msg)
```

```python
# Deserialize msgpack-encoded values
consumer = KafkaConsumer(value_deserializer=msgpack.loads)
consumer.subscribe(['msgpackfoo'])
for msg in consumer:
    assert isinstance(msg.value, dict)
```

```python
# Access record headers. The returned value is a list of tuples
# with str, bytes for key and value
for msg in consumer:
    print(msg.headers)
```

```python
# Read only committed messages from transactional topic
consumer = KafkaConsumer(isolation_level='read_committed')
consumer.subscribe(['txn_topic'])
for msg in consumer:
    print(msg)
```

```python
# Get consumer metrics
metrics = consumer.metrics()
for metric in metrics.items():
    print(metric)
```
